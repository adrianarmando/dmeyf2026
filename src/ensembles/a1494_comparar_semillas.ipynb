### 2.2.5 Asignación de Hiperparámetros y Estimación de AUC
Se define la tabla `tb_nueva` con los hiperparámetros específicos (un solo valor por parámetro) para asignarlos a `PARAM$out$lgbm$mejores_hiperparametros` y continuar el flujo normal.


```r
# En el argumento x llegan los parmaetros de LightGBM
#  devuelve la AUC en cross validation del modelo entrenado

Estimar_AUC_lightgbm <- function(x) {

  # x pisa (o agrega) a param_fijos
  param_completo <- modifyList(PARAM$lgbm$param_fijos, x)

  # entreno LightGBM
  modelocv <- lgb.cv(
    data= dtrain,
    nfold= PARAM$hyperparametertuning$xval_folds,
    stratified= TRUE,
    param= param_completo
  )

  # obtengo la ganancia
  AUC <- modelocv$best_score

  # hago espacio en la memoria
  rm(modelocv)
  gc(full= TRUE, verbose= FALSE)

  message(format(Sys.time(), "%a %b %d %X %Y  "),
    toString(x),
    " AUC ", AUC
  )

  return(AUC)
}
```

```r
# definicion de hiperparametros puntuales para evaluacion
tb_nueva <- CJ(
  learning_rate    = c(0.02),
  num_leaves       = c(50),
  min_data_in_leaf = c(200),
  feature_fraction = c(0.4),
  num_iterations   = c(300)
)
```

```r
# veo que tiene la tabla antes de procesar
tb_nueva
```

```r
# registro a registro calculo la AUC
tb_nueva[, AUC:= Estimar_AUC_lightgbm( list(.SD)),
  by=1:nrow(tb_nueva) ]
```

```r
# veo que tiene la tabla DESPUES de procesar
tb_nueva
```

```r
fwrite( tb_nueva,
  file= "tb_grid_serach_01.txt",
  sep="\t",
  append= TRUE
)
```

```r
setorder( tb_nueva, -AUC)
PARAM$out$lgbm$AUC <- tb_nueva[1, AUC]
PARAM$out$lgbm$mejores_hiperparametros <- as.list( tb_nueva[1] )
PARAM$out$lgbm$mejores_hiperparametros$AUC <- NULL
PARAM$out$lgbm$mejores_hiperparametros
```

```r
# este punto es muy SUTIL  y será revisado en la Clase 05

param_normalizado <- copy(param_final)
param_normalizado$min_data_in_leaf <-  round(param_final$min_data_in_leaf / PARAM$trainingstrategy$undersampling)
```

### Entrenamiento, Scoring y Subida a Kaggle para cada Semilla
Iteramos sobre las 4 semillas definidas en `PARAM$semillas`. Para cada semilla se entrena LightGBM con `param_normalizado`, se guarda la importancia de variables y el modelo, se realiza el scoring sobre `202109` y se envían a Kaggle los archivos para cada corte de envíos incluyendo la semilla y cada hiperparámetro en el mensaje.


```r
# aplico el modelo a los datos sin clase (202109)
dfuture <- dataset[foto_mes == 202109]
dir.create("kaggle", showWarnings = FALSE)

# construyo cadena con todos los hiperparametros para el mensaje de Kaggle
hiperparams_str <- paste(
  names(PARAM$out$lgbm$mejores_hiperparametros),
  PARAM$out$lgbm$mejores_hiperparametros,
  sep = "=",
  collapse = " "
)

cat("Hiperparametros a utilizar:", hiperparams_str, "\n")
cat("Semillas a procesar:", paste(PARAM$semillas, collapse = ", "), "\n\n")

for (semilla_actual in PARAM$semillas) {

  cat("===============================================================\n")
  cat("Iniciando procesamiento de Semilla:", semilla_actual, "\n")
  cat("Fecha/Hora:", format(Sys.time(), "%a %b %d %X %Y"), "\n")
  cat("===============================================================\n")

  # seteo semilla en R y en parametros de LightGBM
  set.seed(semilla_actual, kind = "L'Ecuyer-CMRG")
  param_normalizado$seed <- semilla_actual

  # entreno LightGBM
  cat("Entrenando LightGBM con semilla", semilla_actual, "...\n")
  modelo_final <- lgb.train(
    data = dtrain,
    param = param_normalizado
  )

  # ahora imprimo la importancia de variables
  tb_importancia <- as.data.table(lgb.importance(modelo_final))
  archivo_importancia <- paste0("impo_s", semilla_actual, ".txt")
  fwrite(tb_importancia,
    file = archivo_importancia,
    sep = "\t"
  )

  # grabo a disco el modelo
  lgb.save(modelo_final, paste0("modelo_s", semilla_actual, ".txt"))

  # aplico el modelo a los datos nuevos (202109)
  cat("Generando prediccion sobre 202109...\n")
  prediccion <- predict(
    modelo_final,
    data.matrix(dfuture[, campos_buenos, with = FALSE])
  )

  # tabla de prediccion
  tb_prediccion <- dfuture[, list(numero_de_cliente)]
  tb_prediccion[, prob := prediccion]

  # grabo las probabilidades del modelo
  fwrite(tb_prediccion,
    file = paste0("prediccion_s", semilla_actual, ".txt"),
    sep = "\t"
  )

  # genero archivos con los envios y subo a Kaggle
  setorder(tb_prediccion, -prob)

  for (envios in PARAM$kaggle$cortes) {

    tb_prediccion[, Predicted := 0L] # seteo inicial a 0
    tb_prediccion[1:envios, Predicted := 1L] # marco los primeros envios

    archivo_kaggle <- paste0("./kaggle/KA", PARAM$experimento, "_s", semilla_actual, "_", envios, ".csv")

    # grabo el archivo
    fwrite(tb_prediccion[, list(numero_de_cliente, Predicted)],
      file = archivo_kaggle,
      sep = ","
    )

    # subida a Kaggle, armo la linea de comando
    comando <- "kaggle competitions submit"
    competencia <- paste("-c", PARAM$kaggle$competencia)
    arch <- paste("-f", archivo_kaggle)

    mensaje <- paste0("-m 'envios=", envios,
      "  semilla=", semilla_actual,
      "  ", hiperparams_str,
      "'"
    )

    linea <- paste(comando, competencia, arch, mensaje)
    cat("Ejecutando submit:", linea, "\n")

    salida <- system(linea, intern = TRUE) # el submit a Kaggle
    cat(salida, "\n")
    Sys.sleep(45)
  }
}
```
